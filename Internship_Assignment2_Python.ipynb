```python
#Running sum of 1D array
class Solution:
    def runningSum(self, nums):
        result = []
        current_sum = 0
        for n in nums:
            current_sum += n
            result.append(current_sum)
        return result

arr = input("Enter numbers with spaces")
nums = list(map(int, arr.split()))
s = Solution()
print(s.runningSum(nums))
```

```python
# Shuffle the Array
def shuffle(nums, n):
    result = []
    for i in range(n):
        result.append(nums[i])
        result.append(nums[i + n])
    return result

arr = input("Enter numbers with spaces") #should be an array with even number of elements
nums = list(map(int,arr.split()))
n = int(input("Enter a number")) #Should be half the length of the array
print(shuffle(nums, n))
```

```python
# Kids With the Greatest Number of Candies
def kids_with_candies(candies, extraCandies):
    max_candies = max(candies)
    result = []
    for c in candies:
        result.append(c + extraCandies >= max_candies)
    return result


arr = input("Enter candies for each kid (space-separated): ")
candies = list(map(int, arr.split()))
extra = int(input("Enter extraCandies: "))
print(kids_with_candies(candies, extra))
```
